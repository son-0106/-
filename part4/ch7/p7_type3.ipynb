### 문제1-1

```python
import pandas as pd
# df = pd.read_csv("clam.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/clam.csv")

# 데이터셋 분할
print(df.shape)
train = df.iloc[:210]
test = df.iloc[210:]
print(train.shape)

print(train.head())
```

```python
from statsmodels.formula.api import logit
import numpy as np

# 1) 로지스틱 회귀 모델 생성 및 학습
model = logit("gender ~ weight", data=train).fit()

# 2) 오즈비 계산
odds_ratio = np.exp(model.params['weight'])
print(round(odds_ratio, 4))
```

### 문제1-2

```python
# 1) 로지스틱 회귀 모델 생성 및 학습
formula = "gender ~ age + length + diameter + height + weight"
model = logit(formula, data=train).fit()

# 2) 잔차이탈도
print(round(-2 * model.llf,2))
```

### 문제1-3

```python
from sklearn.metrics import accuracy_score

# 1) test데이터를 사용해 예측 (0.5 미만: 0, 0.5 이상 1)
model = logit("gender ~ weight", data=train).fit()
target = test.pop('gender')
pred = model.predict(test)
pred = (pred > 0.5).astype(int)

# 2) 실제 값과 예측값을 사용하여 정확도 계산
accuracy = accuracy_score(target, pred)

# 3) 오류율 계산
error_rate = 1 - accuracy
print(round(error_rate, 3))
# 0.478
```

###문제1 연습

```python
import pandas as pd
# df = pd.read_csv("clam.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/clam.csv")
from statsmodels.formula.api import logit

train=df.iloc[:210].copy()
test=df.iloc[210:].copy()

#1-1
model_1=logit('gender~weight',data=train).fit()
import numpy as np
odds_ratio=np.exp(model_1.params['weight'])
print(round(odds_ratio,4))

#1-2
model=logit('gender~age+length+diameter+height+weight',data=train).fit()
print(round(-2*model.llf,2))

#1-3
target=test.pop('gender')
pred=model.predict(test)
pred=(pred>0.5).astype(int)

from sklearn.metrics import accuracy_score
accuracy=accuracy_score(target,pred)
print(round(1-accuracy,3))
```

```python
import pandas as pd
# df = pd.read_csv("system_cpu.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/system_cpu.csv")

# 1) ERP와 각 변수 사이의 상관계수 계산
corr_matrix = df.corr()

# 2) ERP와 다른 변수들과의 상관계수 출력
erp_corr = corr_matrix['ERP'].sort_values(ascending=False)
print(erp_corr)
```

### 문제2-2, 2-3

```python
from statsmodels.formula.api import ols

# 1) CPU가 100 미만인 데이터 필터링
filtered_df = df[df['CPU'] < 100]

# 2) 선형회귀 모델 생성: ERP를 종속 변수로, 나머지 변수들을 독립 변수로 설정
model = ols('ERP ~ Feature1 + Feature2 + Feature3 + CPU', data=filtered_df).fit()

# 3) 모델 요약 정보 출력
print(model.summary())
```

###문제2 연습

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/system_cpu.csv")
print(df.head())

#2-1
corr=df.corr()
erp_corr=corr['ERP'].sort_values(ascending=False).iloc[1]
print(round(erp_corr,3))

#2-2
f_df=df[df['CPU']<100]
from statsmodels.formula.api import ols
model=ols('ERP~Feature1+Feature2+Feature3+CPU',data=f_df).fit()
print(model.summary())
print(round(model.rsquared,3))

#2-3
print(round(model.pvalues[1:].max(),3))
```
